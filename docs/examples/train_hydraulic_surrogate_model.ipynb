# Hydraulic Surrogate Model -- Training

This example demonstrates how to create, and train a hydraulic surrogate model.

```python
%pip install epyt-control[hydsurrogate] --quiet
```

```python
from epyt_control.models import PIGNNModel
```

Create a new hydrauic surrogate model for Anytown:

```python
model = PIGNNModel.from_network("anytown")
#model = PIGNNModel("Anytown.inp")  # Alternatively, you can load an arbitrary .inp file via the constructor
```

Generate test data and training demands from the .inp file:

```python
model.load_from_inp()
model.prepare_data(train_ratio=0.6, val_ratio=0.2)
```

Build the neural surrogate model by calling [epyt_control.models.PIGNNModel.build_model](https://epyt-control.readthedocs.io/en/stable/epyt_control.models.html#epyt_control.models.pi_gnn_hydsurrogate.build_model):

```python
model.build_model()
```

Train the model by calling [.train()](https://epyt-control.readthedocs.io/en/stable/epyt_control.models.html#epyt_control.models.pi_gnn_hydsurrogate.train) and store it in a file by calling [.save_model()](https://epyt-control.readthedocs.io/en/stable/epyt_control.models.html#epyt_control.models.pi_gnn_hydsurrogate.save_model):

```python
model.train()
model.save_model(f"anytown_hydsurrogate_pignn.pt")
```

Evaluate its performance on the test data (see [.evaluate](https://epyt-control.readthedocs.io/en/stable/epyt_control.models.html#epyt_control.models.pi_gnn_hydsurrogate.evaluate) function) -- i.e., compare predicted heads and flows to EPANET simulation:

```python
pred = model.evaluate()
```
